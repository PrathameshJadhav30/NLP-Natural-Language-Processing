# Perform pre-processing (Tokenization, Stop word removal and Stemming) of Text

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Download the missing resource

# Sample text
text = "I am Prathamesh Jadhav, a passionate and dedicated student of Artificial Intelligence and Data Science Engineering with a strong interest in solving real-world problems through innovative technology. My academic journey has provided me with a solid foundation in Machine Learning, Data Structures & Algorithms, Cloud Computing, and Web Development. I have hands-on experience in building projects using Java, C++, Python, JavaScript, and the MERN stack, along with HTML/CSS for responsive and modern UI designs.I enjoy competitive programming and problem-solving, with a focus on writing clean, optimized, and well-documented code. My areas of interest extend to Natural Language Processing, Data Analytics, and AI-driven automation. I actively work on improving my skills through projects, coding challenges, and open-source contributions.Alongside technical expertise, I possess strong communication, teamwork, and leadership skills, allowing me to collaborate effectively in both academic and professional environments. I am motivated to apply my knowledge in creating impactful solutions and continuously learning new technologies to stay ahead in the rapidly evolving tech industry."

# Tokenization
sentences = sent_tokenize(text)
words = word_tokenize(text)
print("Sentences:", sentences)
print("Words:", words)

# Stop word removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words]
print("Filtered Words:", filtered_words)

# Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print("Stemmed Words:", stemmed_words)
```

# Student Activity:

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import matplotlib.pyplot as plt

# Download NLTK resources
nltk.download('punkt')
nltk.download('stopwords')

# Sample paragraph (Health domain)
text = """Regular exercise is essential for maintaining good health.
It improves cardiovascular fitness, strengthens muscles, and boosts mental well-being.
Doctors recommend at least 30 minutes of moderate physical activity daily to reduce the risk of chronic diseases."""

# Tokenization
sentences = sent_tokenize(text)
words = word_tokenize(text)

print("Sentences:", sentences)
print("Words:", words)

# Stop word removal
stop_words = set(stopwords.words('english'))
stop_word_count = sum(1 for word in words if word.lower() in stop_words)
filtered_words = [word for word in words if word.lower() not in stop_words]

print("\nFiltered Words (No Stop Words):", filtered_words)
print("Number of Stop Words:", stop_word_count)

# Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
stemmed_word_count = len(stemmed_words)

print("\nStemmed Words:", stemmed_words)
print("Number of Stemmed Words:", stemmed_word_count)

# Visualization
labels = ['Stop Words', 'Stemmed Words']
counts = [stop_word_count, stemmed_word_count]

plt.bar(labels, counts, color=['red', 'green'])
plt.title("Stop Words vs Stemmed Words Count")
plt.ylabel("Count")
plt.show()
```
